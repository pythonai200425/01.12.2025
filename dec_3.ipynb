```python
# MSE case 1 
arr1 = np.array([3, 5, 7])  # true
arr2 = np.array([2, 4, 6])  # predicted
# (3 - 2)² + (5 - 4)² + (7 - 6)²
# -----------------------------
#              3

print('np.mean((arr1 - arr2)**2):     ', np.mean((arr1 - arr2)**2))
```

```python
# MSE case 2
arr = np.array([3, 5, 7])  # predicted
arr_avg = 5  # true
# (5 - 3)² + (5 - 5)² + (5 - 7)²
# -----------------------------
#             3

print('np.mean((arr - arr.mean())**2):', np.mean((arr - arr.mean())**2))
```
